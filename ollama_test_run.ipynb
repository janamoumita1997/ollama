## pass images to generate model

```python
## images should be in base64 format
import base64
import ollama

image_path = "/home/akash/Downloads/graph_rag.png"

with open (image_path,'rb') as f:
    image_bytes = f.read()

image_64 = base64.b64encode(image_bytes).decode("utf-8")

response = ollama.generate(model = "lukaspetrik/gemma3-tools:4b",images=[image_64],prompt="give me a caption for the image.")
print(response.response)
```

## How to manupulate system prompt

```python
response = ollama.generate(model="llama3.2:1b",prompt = "why does moon glow?",system="You are an funny assistance, you explain things in funny way")
print(response.response)
```

```python
## add option inside generate

response = ollama.generate(model="llama3.2:1b",prompt = "why is the ocean blue?",
                options = {
                    "temperature":0.3,
                    "top_p":0.5,
                    "top_k":45
                }).response
print(response)
```
